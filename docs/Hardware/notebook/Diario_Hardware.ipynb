## Detalle pines MEGA 2560 (cerrojo + láser + UART)

- **LED RGB:** 44/45/46 (R/G/B, PWM — feedback local)
- **Keypad 4x4 ROW:** 22/24/26/28
- **Keypad 4x4 COL:** 30/32/34/36
- **Servo MG90S:** 9 (PWM)
- **KY-008 láser S:** 8
- **LDR discreta:** 5V → ● → 7 + 10kΩ → GND (INPUT, divisor en pin 7)
- **UART Gateway:** 16/17 @ 38400 baud (RX2/TX2 hacia ESP32)

Ver `docs/fritzing/AetherNet-P3-MEGA-Cerrojo-v1-breadboard.png` y `docs/fritzing/AetherNet-P5-Laser-v1-breadboard.png`.


````python
import re
from pathlib import Path

# Parsea config.h del MEGA — valida pines / defines trazables
candidates = list(Path(".").rglob("config.h"))
cfg = candidates[0] if candidates else Path("firmware/mega-access/config.h")
print(f"config.h candidato: {cfg} exists={cfg.exists()}")
if cfg.exists():
    text = cfg.read_text(encoding="utf-8", errors="ignore")
    pattern = r"VALID_PIN|LED|LASER|ROW_PINS"
    hits = re.findall(pattern, text)
    print(f"hits VALID_PIN|LED|LASER|ROW_PINS: {len(hits)}", hits[:20])
    # detalle por define
    for line in text.splitlines():
        if re.search(pattern, line):
            print(line.strip())
else:
    print("config.h no encontrado — verifica firmware/mega-access/")
````
